# Imbalanced Binary Classification — Exploratory Analysis

Interactive notebook for data exploration, feature analysis, and model evaluation.
The production pipeline lives in `main.py` and `src/`.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys; sys.path.insert(0, "..")

from src.config import *
from src.data_loader import load_datasets, summarize
from src.preprocessing import build_preprocessor
from src.visualize import *

%matplotlib inline
plt.rcParams["figure.dpi"] = 120
````

## 1 — Load Data

````python
labeled, unlabeled, unlabeled_idx = load_datasets()
summarize(labeled, "Labeled")
````

## 2 — Class Distribution

````python
plot_class_distribution(labeled)
plt.show()
````

## 3 — Missing Values

````python
missing = labeled.isnull().sum()
missing[missing > 0].sort_values(ascending=False).plot(
    kind="barh", color="#ef4444", figsize=(8, 4),
    title="Missing Values by Feature"
)
plt.xlabel("Count")
plt.tight_layout()
plt.show()
````

## 4 — Numerical Feature Distributions

````python
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for i, col in enumerate(NUMERIC_FEATURES):
    ax = axes[i // 4][i % 4]
    labeled[col].hist(bins=30, ax=ax, color="#6366f1", edgecolor="white")
    ax.set_title(col)
if len(NUMERIC_FEATURES) < 8:
    axes[1][3].axis("off")
fig.suptitle("Numerical Feature Distributions", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
````

## 5 — Correlation Matrix

````python
plot_correlation_matrix(labeled)
plt.show()
````

## 6 — Feature vs Target Analysis

````python
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for i, col in enumerate(NUMERIC_FEATURES):
    ax = axes[i // 4][i % 4]
    for label, color in [(0, "#3b82f6"), (1, "#ef4444")]:
        subset = labeled[labeled[TARGET] == label]
        ax.hist(subset[col], bins=25, alpha=0.6, label=f"class {label}", color=color)
    ax.set_title(col)
    ax.legend(fontsize=8)
if len(NUMERIC_FEATURES) < 8:
    axes[1][3].axis("off")
fig.suptitle("Feature Distributions by Class", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
````
